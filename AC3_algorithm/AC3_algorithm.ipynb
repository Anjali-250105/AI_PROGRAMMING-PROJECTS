```python
from collections import deque

# =========================================================
# TELANGANA 33 DISTRICTS - MAP COLORING USING AC-3
# =========================================================

neighbors = {

    "Adilabad": ["Kumuram Bheem", "Nirmal"],

    "Kumuram Bheem": [
        "Adilabad", "Mancherial", "Nirmal"
    ],

    "Mancherial": [
        "Kumuram Bheem", "Nirmal",
        "Peddapalli", "Jayashankar Bhupalpally"
    ],

    "Nirmal": [
        "Adilabad", "Kumuram Bheem", "Mancherial",
        "Nizamabad", "Jagtial", "Kamareddy"
    ],

    "Nizamabad": [
        "Nirmal", "Jagtial", "Kamareddy"
    ],

    "Jagtial": [
        "Nirmal", "Nizamabad", "Peddapalli",
        "Rajanna Sircilla", "Karimnagar"
    ],

    "Peddapalli": [
        "Mancherial", "Jagtial",
        "Karimnagar", "Jayashankar Bhupalpally"
    ],

    "Rajanna Sircilla": [
        "Jagtial", "Karimnagar",
        "Siddipet", "Kamareddy"
    ],

    "Karimnagar": [
        "Jagtial", "Peddapalli",
        "Rajanna Sircilla", "Siddipet"
    ],

    "Kamareddy": [
        "Nizamabad", "Nirmal",
        "Rajanna Sircilla", "Siddipet", "Medak"
    ],

    "Jayashankar Bhupalpally": [
        "Mancherial", "Peddapalli",
        "Mulugu", "Bhadradri Kothagudem",
        "Mahabubabad", "Warangal"
    ],

    "Mulugu": [
        "Jayashankar Bhupalpally",
        "Bhadradri Kothagudem",
        "Mahabubabad"
    ],

    "Bhadradri Kothagudem": [
        "Jayashankar Bhupalpally",
        "Mulugu", "Mahabubabad", "Khammam"
    ],

    "Mahabubabad": [
        "Jayashankar Bhupalpally",
        "Mulugu", "Bhadradri Kothagudem",
        "Khammam", "Suryapet", "Warangal"
    ],

    "Warangal": [
        "Jayashankar Bhupalpally",
        "Mahabubabad",
        "Hanumakonda", "Jangaon"
    ],

    "Hanumakonda": [
        "Warangal", "Jangaon", "Siddipet"
    ],

    "Jangaon": [
        "Warangal", "Hanumakonda",
        "Siddipet", "Yadadri Bhuvanagiri",
        "Nalgonda"
    ],

    "Siddipet": [
        "Nirmal", "Rajanna Sircilla",
        "Karimnagar", "Kamareddy",
        "Medak", "Yadadri Bhuvanagiri",
        "Jangaon", "Hanumakonda"
    ],

    "Medak": [
        "Kamareddy", "Siddipet", "Sangareddy"
    ],

    "Sangareddy": [
        "Medak", "Vikarabad",
        "Rangareddy", "Medchal-Malkajgiri"
    ],

    "Medchal-Malkajgiri": [
        "Sangareddy", "Rangareddy",
        "Hyderabad", "Yadadri Bhuvanagiri"
    ],

    "Hyderabad": [
        "Medchal-Malkajgiri", "Rangareddy"
    ],

    "Rangareddy": [
        "Sangareddy", "Vikarabad",
        "Medchal-Malkajgiri", "Hyderabad",
        "Mahabubnagar", "Nalgonda"
    ],

    "Vikarabad": [
        "Sangareddy", "Rangareddy",
        "Mahabubnagar", "Narayanpet"
    ],

    "Mahabubnagar": [
        "Vikarabad", "Rangareddy",
        "Narayanpet", "Wanaparthy",
        "Nagarkurnool"
    ],

    "Narayanpet": [
        "Vikarabad", "Mahabubnagar",
        "Wanaparthy", "Jogulamba Gadwal"
    ],

    "Wanaparthy": [
        "Narayanpet", "Mahabubnagar",
        "Nagarkurnool", "Jogulamba Gadwal"
    ],

    "Jogulamba Gadwal": [
        "Narayanpet", "Wanaparthy",
        "Nagarkurnool"
    ],

    "Nagarkurnool": [
        "Mahabubnagar", "Wanaparthy",
        "Jogulamba Gadwal", "Nalgonda", "Suryapet"
    ],

    "Nalgonda": [
        "Jangaon", "Rangareddy",
        "Nagarkurnool", "Suryapet",
        "Yadadri Bhuvanagiri"
    ],

    "Suryapet": [
        "Mahabubabad", "Nalgonda",
        "Nagarkurnool", "Khammam"
    ],

    "Khammam": [
        "Bhadradri Kothagudem",
        "Mahabubabad", "Suryapet"
    ],

    "Yadadri Bhuvanagiri": [
        "Siddipet", "Jangaon",
        "Medchal-Malkajgiri", "Nalgonda"
    ]
}


# =========================================================
# MAKE GRAPH SYMMETRIC
# =========================================================

for district in list(neighbors):

    for neighbour in neighbors[district]:

        if district not in neighbors[neighbour]:
            neighbors[neighbour].append(district)


# =========================================================
# COLORS / DOMAINS
# =========================================================

colors = ["Red", "Green", "Blue", "Yellow"]

domains = {
    district: colors.copy()
    for district in neighbors
}


# =========================================================
# REVISE FUNCTION
# =========================================================

def revise(domains, district1, district2):

    revised = False

    for color in domains[district1].copy():

        # Check if district2 has at least
        # one different color

        if not any(
            color != other_color
            for other_color in domains[district2]
        ):

            domains[district1].remove(color)

            revised = True

    return revised


# =========================================================
# AC-3 ALGORITHM
# =========================================================

def ac3(domains):

    queue = deque()

    # Add all arcs to queue

    for district in neighbors:

        for neighbour in neighbors[district]:

            queue.append(
                (district, neighbour)
            )


    while queue:

        district1, district2 = queue.popleft()

        if revise(
            domains,
            district1,
            district2
        ):

            # No possible color

            if len(domains[district1]) == 0:
                return False


            # Recheck affected neighbours

            for neighbour in neighbors[district1]:

                if neighbour != district2:

                    queue.append(
                        (neighbour, district1)
                    )

    return True


# =========================================================
# BACKTRACKING
# =========================================================

def backtracking(domains):

    # All districts assigned

    if all(
        len(domains[district]) == 1
        for district in domains
    ):

        return domains


    # Choose district with smallest domain

    district = min(
        (
            d for d in domains
            if len(domains[d]) > 1
        ),
        key=lambda d: len(domains[d])
    )


    # Try every available color

    for color in domains[district]:

        new_domains = {
            d: domains[d].copy()
            for d in domains
        }

        new_domains[district] = [color]


        # Apply AC-3

        if ac3(new_domains):

            result = backtracking(new_domains)

            if result is not None:
                return result


    return None


# =========================================================
# SOLVE
# =========================================================

print("Running AC-3...")

if ac3(domains):

    solution = backtracking(domains)

else:

    solution = None


# =========================================================
# FINAL OUTPUT
# =========================================================

if solution:

    print("\n======================================")
    print("TELANGANA MAP COLORING USING AC-3")
    print("======================================\n")

    for district in sorted(solution):

        print(
            f"{district:30} -> "
            f"{solution[district][0]}"
        )

else:

    print("No solution exists.")


# =========================================================
# CHECK CONSTRAINTS
# =========================================================

print("\nChecking constraints...")

valid = True

for district in neighbors:

    for neighbour in neighbors[district]:

        if (
            solution[district][0]
            == solution[neighbour][0]
        ):

            print(
                "Conflict:",
                district,
                "-",
                neighbour
            )

            valid = False


if valid:

    print("All constraints satisfied!")

else:

    print("Some constraints are violated.")
```
